```python
import numpy as np

# 1. Base CAPEX from Aspen (USD)
base_capex = 5320000 

# 2. Extracted Reactor Volumes from Excel (m^3)
v_rwgs = 0.026770101

v_ft = 0.064599647


# 3. Industry Reference Data (0.1 m^3 reactor costs $50,000)
v_ref = 0.1
c_ref = 50000

# 4. Six-Tenths Scaling Function
def calculate_scaled_cost(v_new, v_ref, c_ref):
    return c_ref * ((v_new / v_ref) ** 0.6)

# 5. Execute Calculations
cost_rwgs = calculate_scaled_cost(v_rwgs, v_ref, c_ref)
cost_ft = calculate_scaled_cost(v_ft, v_ref, c_ref)

# 6. Final Plant CAPEX
total_capex = base_capex + cost_rwgs + cost_ft

# 7. Print Results formatting
print(f"Estimated RWGS Reactor Cost: ${cost_rwgs:,.2f}")
print(f"Estimated FT Reactor Cost:   ${cost_ft:,.2f}")
print(f"---------------------------------------------")
print(f"Total Adjusted Plant CAPEX:  ${total_capex:,.2f}")
```

```python
# 1. Annual Production (kg/year)
hourly_yield = 31.31  # kg/hr from Aspen optimization
operating_hours = 8000
annual_production = hourly_yield * operating_hours

# 2. Annualize the CAPEX (10% discount rate over 20 years)
discount_rate = 0.10
plant_life = 20
crf = (discount_rate * (1 + discount_rate)**plant_life) / (((1 + discount_rate)**plant_life) - 1)
annualized_capex = total_capex * crf

# 3. Total Annual Cost
annual_opex = 1900000  # $1.90M from Aspen Excel extraction
total_annual_cost = annualized_capex + annual_opex

# 4. Levelized Cost of Fuel (USD/kg)
lcof = total_annual_cost / annual_production

# 5. Output
print(f"Annual Jet Fuel Production: {annual_production:,.0f} kg")
print(f"Annualized CAPEX Payment:   ${annualized_capex:,.2f}")
print(f"Total Annual Cost:          ${total_annual_cost:,.2f}")
print(f"---------------------------------------------")
print(f"Levelized Cost of Jet Fuel: ${lcof:,.2f} per kg")
```
