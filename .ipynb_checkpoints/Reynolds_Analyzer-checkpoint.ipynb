```python
#Week 2|Thursday| 5th May 2026
#=======================================================================================================================================================
#PROJECT:Automated Reynolds &Friction Factor Analyzer
#MODULE:Fluid Regime Classification Engine
#PHYSICAL LAW:Momentum Balance/Inertial vs Viscous Forces
#========================================================================================================================================================
print('=======================================================================================================================================================')
print(f'\n{"\033[4;50m"}AUTOMATED PIPING FLUID REGIME & FRICTION SOLVER.{"\033[0m"}')
#1.Capture physical parameters with explicit SI units in the prompt.
rho=float(input('\nEnter fluid Density (rho) in [kg/m3]:'))
v=float(input('Enter the viscosity of the fluid(v) in [m/s]:'))
D=float(input('Enter inner pipe diameter(D) in [meters]:'))
mu=float(input('Enter the dynamic viscosity of the fluid(mu) in [Pa.s or kg/m.s]:'))

#2.Check Validation and Compute the Dimensionless Renolds Number
#Parentheses ensure the numerator evaluates fully before division
if rho<=0 or v<=0 or D<=0 or mu<=0:
    print(f'\n{"\033[31m"}[CRITICAL ERROR]:Input Values can not be less than or equal to zero.{"\033[0m"}')
    print(f'{"\033[32m"}Calculation Safely aborted to prevent unphysical matrics.{"\033[0m"}')
else:
    print(f'\n{"\033[32m"}[SUCCESS]:Inputs Validated. Running core engine...{"\033[0m"}')
    Re=(rho*v*D)/mu
    print(f'\n{"\033[4;50m"}RESULTS{"\033[0m"}')
    print(f'\nCalculated Renolds Number(Re):{Re:.2f}')
#3. Multi- Branch Decision Tree to evaluate the Physical Regime
    if Re<=2100:
       print(f'Physical Regime:{"\033[4;50m"}LAMINAR FLOW{"\033[0m"}')
# For laminar flow the friction  factor depends on only Re 
       f=64.0/Re
       print(f'Darcy Friction Factor(f):{f:.4f}')
    elif Re > 2100 and Re <=4000:
         print(f'Physical Regime:{"\033[4;50m"}TRANSITIONAL FLOW{"\033[0m"}')
         print(f'{"\033[32m"}Warning: Flow is physically unstable and mathematically unpredictable.{"\033[0m"}')
         f=0.0 #Placeholder since no single standard equation fits this zone easily
         print('Darcy Friction Factor(f): Cannot evaluate explicitly')
    else:
        print(f'Physical Regime:{"\033[4;50m"}TURBULENT FLOW{"\033[0m"}')
        print(f'{"\033[32m"}Flow is chaotic. Wall roughness constraints must be evaluated next.Accessing Swamee-Jain engine...{"\033[0m"}')
#======================================================================================================================================================
#MODULE:Complete Fluid Regime & Swamee-Jain Solver Engine
#PHYSICAL LAW: Conservation of Momentum % Moody roughness Drag
#======================================================================================================================================================
        import math as m #importing the required library for advanced transcendental function like log10
# Capture the additional physial property required for turbulent flow
        epsilon=float(input('Enter Pipe Absolute Roughness(epislon) in [meter]:'))
# Breaking the Swamee-Jain Equation
        inner_term=(epsilon/(3.7*D))+(5.74/(Re**0.9))
        denominator=(m.log10(inner_term))**2
        f=0.25/denominator
        print(f'\n{"\033[4;50m"}TURBULENT SOLVER OUTPUT{"\033[0m"}')
        print(f'Darcy Friction Factor(f):{f:.4f}')
print('=======================================================================================================================================================')

        
        

          
```
