```python
#========================================================================================================================================================
#WEEK:3  |11th June 2026| 7:05pm
#PRPJECT: Automated VLE Flash Drum Calculator
#MODULE: Numerical Iteration (Interval Bisection Engine)
#PHYSICAL LAW: Rachford-Rice Thermodynamic Phase Equilibrium
#========================================================================================================================================================
print(f'{"\033[4;50m"}THERMODYNAMIC FLASH ENGINE{"\033[0m"}')
#1. Define Feed compositions and Equilibrium Volatilities(K)
ZA=0.30 # Mole Fraction of Component A in Feed
ZB=0.40 # Mole Fraction of Component B in Feed
ZC=0.3 #  Mole Fraction of Component C in Feed
KA=3.5 # Equilibrium Ratio/Constant for Component A(Highly Volatile)
KB=1.2 # Equilibrium Ratio/Constant for Component b(Low Volatility)
KC=0.15 # Equilibrium Ratio/Constant for Component b(Very Low Volatility)
#2. Initialize Numerical Bisection Search Boundaries 
V_low=0.0
V_high=1.0
tolerance=1e-6
error=1.0 # Set initial error high to force entry into the loop iteration = 0
#3. The iterative Numerical Solver Loop
iterations=0
while error>tolerance:
    iterations += 1 # Track how many guesses the computer makes
# Calculate the midpoint guess for the vapor fraction
    V_guess=(V_low+V_high)/2.0
# Evaluating the Rachford-Rice Equation for both components using the guess 
    term_A=(ZA*(KA-1.0))/(1.0+V_guess*(KA-1.0))
    term_B=(ZB*(KB-1.0))/(1.0+V_guess*(KB-1.0))  #This continues for n number of components
    term_C=(ZC*(KC-1.0))/(1.0+V_guess*(KC-1.0))
    f_V= term_A+term_B+term_C
# Checking the physical direction of the error and the shift the boundaries
    if f_V > 0.0:
# Our Vapor fraction guess was too low:move the lower limit up 
       V_low=V_guess
    else:
# Our Vapor fraction guess was too high:move the upper limit down
        V_high=V_guess
    error=abs(V_high-V_low)
# LOOP BROKEN: CONVERGENCE ACHIEVED
print('\nTHERMODYNAMIC CONVERGENCE ACHIEVED')
print(f'\nTotal Iterations executed:{iterations}')
print(f'Final Vapor Fraction (V/F):{V_guess:.4f}')
print(f'Final Liquid Fraction (L/F):{1.0-V_guess:.4f}')
        
```
