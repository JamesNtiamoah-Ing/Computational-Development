```python
#Building a multi-unit mass balance 
#Imagine a seperation train with two units in series seperating a mixture of three hydrocarbon components: X,Y,Z
#We calculate this using 3 independent equations
#Overall Mass Balnce F=D1+D2+B2 (depending on how the question was placed) 
#Component Balance: Balance for component X (F.xF=D1.xD1+D2.xD2+B2.xB2)
#Component Balance for Y: F.yF=D1.yD1+D2.yD2+B2.yB2
#Assuming the laboratory gas chromatograph returns the following mass fractions for our streams:
#Feed(F):X=0.3,Y=0.4,Z=0.3
#Stream D1:X=0.8,Y=0.15,Z=0.05
#Stream D2:X=0.1,Y=0.7,Z=0.2
#Stream B2:X=0.01,Y=0.1,Z=0.89

#===============================================================================================================================================================
# PROJECT: Multi-Unit Mass Balance Linear Solver (Week 1)
# MODULE: Numpy Matrix Engine
# PHYSICAL LAW: Multi-Component Material Conservation(Ax=B)
#========================================================================================================================================================
import numpy as np
print('---MULTI-UNIT SEPARATION TRAIN SOLVER---')
# Defining the matrixes
# DYNAMICALLY CAPTURE THE COEFFICIENT MATRIX A
print('---Input the 3x3 coefficient Matrix A---')
print('Enter the row values(eg.mass fraction) as prompted:')
matrix_list =[]
for row in range(3):
    row_list=[]
    for col in range(3):
# Dynamic prompt showing the exact index we are building
        val=float(input(f'Enter coefficient for Position ({row},{col}):'))
        row_list.append(val)
    matrix_list.append(row_list)
# Convert the raw nested list into a structural array
A=np.array(matrix_list)
print(f'Matrix A Shape:{A.shape}')
#  A=np.array([
#   [1.0,1.0,1.0],#Total mass balance,
#   [0.15,0.70,0.10],#Component X mass fractions
#   [0.15,0.70,0.10]#Component Y mass fractions
#          ])
#=======================================================================================================================================================
# Week 1| Day 3| Saturday| 30th May 2026| 4:15PM
# MODULE:PART 2 -Matrix Structural Validation Loop 
# PHYSICAL PRINCIPLE: System Solvability & Sensor Independence Check
#========================================================================================================================================================
#Compute the system matrix to chek if the matrix can be inverted 
det_A=np.linalg.det(A)
print(f'\n Evaluating System Determinant:{det_A:.4f}')
#Check if the determinant is close to zero ( Singular Matrix Check )
if np.isclose(det_A,0.0):
    print('\n [CRITICAL SENSOR FAILURE]: Singular Matrix Encountered.')
    print('The balance equations are dependent or the sensor is duplicating data.')
    print(' Calculation Safetly aborted to prevent a control system crash')
else:
    print('[SUCESS]: Matrix Structure Validated. Executing Solver Engine...')
#This line only runs if the matrix is healty and invertible 
# Capturing the values of vector B
    F=float(input('Enter the mass flow rate of Feed in kg/hr'))
    xF=float(input('Enter the mass fraction of Component X in Feed [0-1]'))
    yF=float(input('Enter the mass fraction of Component Y in Feed [0-1]'))
    B=np.array([F,F*xF,F*yF])
    print('\n Matrix A (System Coefficients):')
    print(A)
    print('\n Matrix B (Boundary Constants):')
    print(B)

#Numpy Computing the unknown flow rates 
    x=np.linalg.solve(A,B)
    print('\n === VECTORIZED SOLVER OUTPUT===')
    print(f'Distillate 1 Flow Rate (D1):{x[0]:.2f} kg/hr')
    print(f'Distillate 2 Flow Rate (D2):{x[1]:.2f} kg/hr')
    print(f'Bottom 2 Flow Rate (B2):{x[2]:.2f} kg/hr')      
```
