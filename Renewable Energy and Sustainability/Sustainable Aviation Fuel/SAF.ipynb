```python
import numpy as np

# SCENARIO 1: PILOT-SCALE BASELINE (CORRECTED WITH FINAL ANALYSIS RESULTS)
# =====================================================================

# 1. Base CAPEX from Image_rtbN0A (5.37E+06) & Reactor Sizing from Image_jVqBpK
base_capex = 5370000  # Exact analysis output value
v_rwgs = 0.027682494  # Sized RWGS volume (m3)
v_ft = 0.066601347    # Sized FT volume (m3)
v_ref, c_ref = 0.1, 50000 

cost_rwgs = c_ref * ((v_rwgs / v_ref) ** 0.6)
cost_ft = c_ref * ((v_ft / v_ref) ** 0.6)
pilot_capex = base_capex + cost_rwgs + cost_ft

# 2. Production Profile
pilot_yield_hr = 31.3155  
operating_hours = 8000     
pilot_annual_production = pilot_yield_hr * operating_hours

# 3. Capital Recovery Factor (10% over 20 years)
discount_rate = 0.10
plant_life = 20
crf = (discount_rate * (1 + discount_rate)**plant_life) / (((1 + discount_rate)**plant_life) - 1)
pilot_annualized_capex = pilot_capex * crf

# 4. Total Annual Pilot OPEX from Image_rtbN0A (1.77E+06)
pilot_opex = 1770000  # Exact analysis output value
pilot_total_annual_cost = pilot_annualized_capex + pilot_opex

# 5. Pilot LCOF Calculation
pilot_lcof = pilot_total_annual_cost / pilot_annual_production

print("--- SCENARIO 1: PILOT-SCALE BASELINE ---")
print(f"Annual Production:     {pilot_annual_production:,.2f} kg")
print(f"Adjusted CAPEX:        ${pilot_capex:,.2f}")
print(f"Annual OPEX:           ${pilot_opex:,.2f}")
print(f"Pilot LCOF:            ${pilot_lcof:,.2f} per kg\n")


# =====================================================================
# SCENARIO 2: COMMERCIAL-SCALE PROJECTION (100x Scale-Up Verification)
# =====================================================================

# 1. Commercial Scaling Multipliers
scale_factor = 100
scaling_multiplier = scale_factor ** 0.6

# 2. Scale Production Volume
commercial_annual_production = pilot_annual_production * scale_factor

# 3. Scale Capital Assets
commercial_capex = pilot_capex * scaling_multiplier
commercial_annualized_capex = commercial_capex * crf

# 4. Scale Operating Expenditures (Isolating Linear Process Utilities)
pilot_utility_cost = 43716  # Active utilities from your sizing sheet
pilot_fixed_opex = pilot_opex - pilot_utility_cost 

commercial_fixed_opex = pilot_fixed_opex * scaling_multiplier
commercial_utility_cost = pilot_utility_cost * scale_factor
commercial_opex = commercial_fixed_opex + commercial_utility_cost

# 5. Commercial LCOF Metric
commercial_total_annual_cost = commercial_annualized_capex + commercial_opex
commercial_lcof = commercial_total_annual_cost / commercial_annual_production

print("--- SCENARIO 2: COMMERCIAL-SCALE (100x) ---")
print(f"Annual Production:     {commercial_annual_production:,.2f} kg")
print(f"Projected CAPEX:       ${commercial_capex:,.2f}")
print(f"Projected Annual OPEX: ${commercial_opex:,.2f}")
print(f"Commercial LCOF:       ${commercial_lcof:,.2f} per kg")
```
